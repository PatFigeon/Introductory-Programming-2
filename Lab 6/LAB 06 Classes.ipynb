What else can your Class be programmed to do?

*(Oh the `IPython.display` library has a function called `clear_output` which will clear the old output so that you can overlay the new output and give the impression of movement)*

````python
# roguelike Room Objects - write your code here

from IPython.display import clear_output

class Room():
    def __init__(self, width, height):
        self.roomXLength = width
        self.roomYLength = height
        self.playerX = 0
        self.playerY = 0

    def drawBoundaryLine(self):
        print("+" + self.roomXLength * "-" + "+")

    def drawEmptyLine(self):
        print("|" + self.roomXLength * "." + "|")

    def drawPlayerLine(self):
        print("|", end="") # forces no newline
        for columns in range(self.playerX):
            print(".", end="")
        print("@", end="") # draw player
        for columns in range(self.roomXLength - self.playerX - 1): # draw floor after player
            print(".", end="")
        print("|")

    def draw(self):
        self.drawBoundaryLine()
        for rows in range(self.playerY): # draw rows above player
            self.drawEmptyLine()
        self.drawPlayerLine()
        for rows in range(self.roomYLength - self.playerY - 1): # draw rows below player
            self.drawEmptyLine()
        self.drawBoundaryLine()


    def left(self):
        if self.playerX > 0:
            self.playerX = self.playerX - 1
        else:
            print("Invalid move, try again")

    def right(self):
        if self.playerX < self.roomXLength - 1:
            self.playerX = self.playerX + 1
        else:
            print("Invalid move, try again")

    def up(self):
        if self.playerY > 0:
            self.playerY = self.playerY - 1
        else:
            print("Invalid move, try again")

    def down(self):
        if self.playerY < self.roomYLength - 1:
            self.playerY = self.playerY + 1
        else:
            print("Invalid move, try again")
    

myRoom = Room(4,4)
myRoom.draw()

while True:
    s = input()
    if s=='a': myRoom.left()
    if s=='s': myRoom.right()
    if s=='q': myRoom.up()
    if s=='z': myRoom.down()
    if s=='x': 
        print("all done")
        break
    clear_output()
    myRoom.draw()
````
