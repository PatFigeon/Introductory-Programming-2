# Numerical Integration

Write a program to calculate the integral of a function using both the trapezoidal rule and the rectangular rule. That is, we want to calculate an approximation to
\begin{equation}
I = \int_a^b f(x)dx
\end{equation}
which is the area under the curve as shown below
<img src="integralab.png" width="200"/>

We can usually approximate integrals by numerical techniques, by splitting the the area up into slices and then adding up the slices. More, thinner slices, is equivalent to a closer approximation, but is also more work when doing the approximation by hand. Automating this is a perfect job for a computer.

You should write a program that first defines the the limits of the integral (i.e. the values a and b in the equation) and the number of slices / integral divisions. Please don't simply use a library function to compute these integrals - we want to see you implement the algorithms for both the rectangular and trapezoidal rule (explained below).

Your program should calculate the integrals of $f(x) = x$ , $f(x) = sinx$ and $f(x) = e^{-x} sinx$ between $a=0$ and $b=1$ using both rules. You should output the value of the approximation for $n=2$ (two slices) up to $n=100$, and graph the result at each stage, showing how the answer approaches the correct value.

### Rectangular Rule
This works by dividing the area under the curve into rectangular slices as shown below.

<img src="integral1.png" width="200"/>

Let’s take an example of how to apply this to calculate the integral 􏰄$\int_0^1 x^3dx$ for $n = 2$. We first calculate the step size $h = (b − a)/n = (1 − 0)/2 = 0.5$. We now add the two rectangles to get an approximate integral 

\begin{equation}
I = f(x_1)h+f(x_2)h = f(0) \times 0.5 + f(0.5) \times 0.5
\end{equation}

\begin{equation}
= 0 \times 0.5+0.125 \times 0.5 = 0.0625
\end{equation}


We can also do the same for $n=4$, in the figure below $h=(1-0)/4 = 0.25$

<img src="integral2.png" width="200"/>

and the approximate integral is found from adding up 4 slices

\begin{equation}
I = f(x_1)h+f(x_2)h+f(x_3)h+f(x_4)h 
\end{equation}

\begin{equation}
= f(0) \times 0.25 + f(0.25) \times 0.25 + f(0.5) \times 0.25+ f(0.75) \times 0.25
\end{equation}

\begin{equation}
= 0 \times 0.25 + 0.015625 \times 0.25 + 0.125 \times 0.25 + 0.421875 \times 0.25 = 0.140625
\end{equation}

If we were to work this integral out by hand, we would see that the exact answer is 0.25. As we increase $n$ we are at least getting closer to 0.25, although the approximation isn't great.


### Trapeziodal Rule

The trapezoidal rule approximates the integral by constructing **trapezoids** under the function, as shown below. It is clear that this provides a more accurate approximation of the integral.

<img src="integral3.png" width="200"/>

Let’s take an example of how to apply this to calculate the integral 􏰄$\int_0^1 x^3dx$ for $n = 2$. As above, we calculate the step size $h = (b − a)/n = (1 − 0)/2 = 0.5$. We now add the two *trapeziods* to get an approximate integral (double check the formula to ensure that you see why it works)

\begin{equation}
I = \frac{h}{2}\left[{f(x_1)+2f(x_2)+f(x_3)}\right]
\end{equation}

\begin{equation}
= \frac{0.5}{2}\left[{f(0)+2f(0.5)+f(1.0)}\right] = 0.3125
\end{equation}


We can also do the same for $n=4$, in the figure below $h=(1-0)/4 = 0.25$

<img src="integral2.png" width="200"/>

and the approximate integral is found from adding up 4 slices

\begin{equation}
I = \frac{h}{2}\left[{f(x_1)+2f(x_2)+2f(x_3)+2f(x_4)+f(x_5)}\right]
\end{equation}

\begin{equation}
= \frac{0.25}{2}\left[{f(0)+2f(0.25)+2f(0.5)+2f(0.75)+f(1.0)}\right]
\end{equation}

\begin{equation}
= 0.125\left[{0+ 2\times 0.015625 + 2 \times 0.125 + 2 \times 0.421875 +1}\right] = 0.265625
\end{equation}

Obviouilsy after only 4 slices, we are already closely approaching the correct answer.

### Testing

The obvious way to test your program is to compare your results against integrals you have performed by hand. For example, in the results shown above we know the exact result to be 0.25 and we can see that they are converging to the exact result as $n$ is increased.

```python
"""
h = (b-a)/n

I ~= f(x)h
I ~= h/2[f(x_1)+ 2f(x_2) + 2f(x_3)+...+ f(x_n)]
"""

# Todo: graphing

import numpy as np

class Functions:

    def LinearFunction(xArg): # linear function y = x
        functionOutput = xArg
        return functionOutput

    def SineFuntion(xArg): # basic sine function y = sin(x)
        functionOutput = np.sin(xArg)
        return functionOutput

    def ExponentFunction(xArg): # exponential and sine y=e^-x * sin(x)
        functionOutput = np.exp(-xArg)*np.sin(xArg)
        return functionOutput

    functionList = {"line": LinearFunction, "sine": SineFuntion, "expo": ExponentFunction} # store functions in a dict

class Integration:

    def __init__(self, upperLimit, lowerLimit, repetitions):
        self.upperLimit = upperLimit
        self.lowerLimit = lowerLimit
        self.repetitions = repetitions
        self.stepSize = (self.upperLimit - self.lowerLimit)/self.repetitions # h value
        self.totalIntegralValue = 0

    def RectangleRuleIntegral(self):
        # this is an implementation of left rectangle rule; midpoint rectangles would be more accurate, but the brief uses left rectangles
        if selectedFunction != None:
            for i in range(self.repetitions):
                additionValue = selectedFunction(self.lowerLimit + self.stepSize * i) * self.stepSize
                self.totalIntegralValue += additionValue
            return self.totalIntegralValue
        else:
            return -1 # basic error catch: if it is -1, something has gone wrong - in theory this should never trigger, but you never know

    def TrapeziumRuleIntegral(self):
        if selectedFunction != None:
            self.totalIntegralValue = selectedFunction(self.lowerLimit) # initial step - resets counter to first step of this function
            for i in range(self.repetitions - 1): # account for initial at x_0 and final step
                additionValue = 2 * selectedFunction(self.lowerLimit + self.stepSize * (i + 1)) # get value of function at x
                self.totalIntegralValue += additionValue # add to counter
            additionValue = selectedFunction(self.upperLimit) # final step at x_n
            self.totalIntegralValue = (self.stepSize / 2) * (self.totalIntegralValue + additionValue)
            return self.totalIntegralValue
        else:
            return -1

class Setup:

    def FunctionSelect():
            while True:
                try:
                    userSelection = int(input("Select which function you want to integrate:\n1) y=x\n2) y=sin(x)\n3) y=exp(-x) * sin(x)"))
                    if userSelection > 3 or userSelection < 1: # learn to read guys
                        print("There's only 3 functions here, try again.")
                        continue
                    else:
                        break
                # add type validation for brownie points
                except ValueError: # the error is, as usual, behind the keyboard
                    print("Not an int, try that again")
                    continue
                
            match userSelection: # switch, but different because python
                case 1:
                    selectedFunction = Functions.functionList["line"]
                case 2:
                    selectedFunction = Functions.functionList["sine"]
                case 3:
                    selectedFunction = Functions.functionList["expo"]
                case _:
                    selectedFunction = None
            return selectedFunction

    # dynamic limits
    # scope creep is defined as...
    def GetLimits():
        # available function have no discontinuities, so all limits are valid
        limit = []
        while True:
            lowerLimit = input("What is your lower limit of the integral?")
            try:
                lowerLimit = float(lowerLimit) # try to convert input into a real number
                limit.append(lowerLimit) # if it is a number, add it to the array
                break
            except ValueError:
                print("I need a number of some kind.")
                continue

        while True:
            upperLimit = input("What is your upper limit of the integral?")
            try:
                upperLimit = float(upperLimit)
                if upperLimit < lowerLimit: # catch limit discrepancy
                    print("The upper limit must be greater than the lower limit.")
                else:
                    limit.append(upperLimit)
                    break
            except ValueError:
                print("I need a number of some kind.")
                continue
        return limit

    def GetNumberOfIterations():
        while True:
            try:
                iterations = int(input("How many iterations of each rule should I do?"))
                if iterations < 1:
                    print("I can't really do it 0 times or negative times...")
                    continue
                else:
                    break
            except ValueError:
                print("Not an int, try again")
                continue
        return iterations

setup = Setup()
selectedFunction = setup.FunctionSelect()
limitArray = setup.GetLimits()
integral = Integration(limitArray[1], limitArray[0], setup.GetNumberOfIterations())

answer = integral.RectangleRuleIntegral()
print("The rectangle rule gives: ",answer)
answer = integral.TrapeziumRuleIntegral()
print("The trapezium rule gives: ",answer)
```
