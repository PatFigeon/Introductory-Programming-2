````python
# Write your set-up and calculations for question 1
"""
Graph of instantaneous component velocities
"""
g = 9.81
angle = 25
initialV = 7.5
rad = np.deg2rad(angle)

Vx = np.round(initialV*np.cos(rad), 2)
Vy = np.round(initialV*np.sin(rad), 2)


x = np.linspace(0, 0.65, num=10) # x is functionally time
velocity_x = Vx+(0*x) # no change as no drag
"""
No upwards acceleration, thus immediately decelerating

By UVAST: v = u+at
"""
velocity_y = Vy-g*x

timeInAir = np.round(2*Vy/g, 2)


````

````python
# Write your calculations for question 2 here

velocity_mag = np.sqrt(velocity_x**2 + velocity_y**2)

position_x = Vx*x
position_y = Vy*x + 0.5*-g*x**2


````

````python
# Make your plots here
plt.figure(figsize=[10,10])

# instantaneous velocity
plt.subplot(2,2,1)
plt.plot(x, velocity_x, color = 'b', label='Instantaneous x velocity')
plt.plot(x, velocity_y, color = 'r', label='Instantaneous y velocity')
plt.xlabel('Time (s)')
plt.ylabel('Velocity (m/s)')
plt.title('Instantaneous component velocty against time')

# magnitude of velocity
plt.subplot(2,2,2)
plt.plot(x, velocity_mag, color = 'g', label='Magnitude of velocity')
plt.xlabel('Time (s)')
plt.ylabel('Velocity (m/s)')
plt.title('Magnitude of velocity against time')

# instantaneous position
plt.subplot(2,2,3)
plt.plot(x, position_x, color = 'b', label='Horizontal position')
plt.plot(x, position_y, color = 'r', label='Vertical Altitude')
plt.xlabel('Time (s)')
plt.ylabel('Distance (m)')
plt.title('Instantaneous component distances against time')

# altitude against horizontal distance
plt.subplot(2,2,4)
plt.plot(position_x, position_y, color = 'm', label='Altitude against horizontal distance')
plt.xlabel('Horizontal displacement (m)')
plt.ylabel('Altitude (m)')
plt.title('Altitude against horizontal distance')



plt.legend() # Shows the plot legend
plt.tight_layout()
plt.show()
plt.savefig('projectile_motion')
````
